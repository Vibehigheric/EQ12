## 3. Responses API Implementation

Implement reasoning models using the Responses API, showing the difference from Chat Completions API and demonstrating improved performance.

```python
# Responses API vs Chat Completions API comparison
class ResponsesAPIDemo:
    """
    Demonstrate the Responses API advantages over Chat Completions API
    """

    def __init__(self):
        self.client = client
        logger.info("Initialized ResponsesAPIDemo")

    def compare_apis(self, prompt: str) -> dict[str, Any]:
        """
        Compare Responses API vs Chat Completions API performance
        """
        results = {}

        # EQ12 Cannabis Compliance Analysis Problem
        cannabis_prompt = """
        Analyze this cannabis compliance scenario for EQ12:

        A dispensary in California wants to expand to Nevada and Colorado.
        They need to understand:
        1. Licensing requirements across all three states
        2. Product testing and labeling differences
        3. Tax implications and reporting requirements
        4. Interstate commerce restrictions
        5. Banking compliance considerations

        Create a comprehensive compliance roadmap with risk assessment.
        """

        print("🔄 COMPARING APIS...")

        # Test 1: Responses API (Recommended for reasoning models)
        try:
            start_time = time.time()

            responses_result = self.client.responses.create(
                model="gpt-5",
                reasoning={"effort": "medium"},
                instructions="You are an expert cannabis compliance consultant for EQ12.",
                input=[{"role": "user", "content": cannabis_prompt}],
            )

            responses_time = time.time() - start_time

            results["responses_api"] = {
                "response": responses_result,
                "execution_time": responses_time,
                "token_count": responses_result.usage.total_tokens,
                "reasoning_tokens": responses_result.usage.output_tokens_details.get(
                    "reasoning_tokens", 0
                ),
                "output_length": len(responses_result.output_text),
            }

            print("✅ Responses API completed")

        except Exception as e:
            logger.error(f"Responses API error: {e}")
            results["responses_api"] = {"error": str(e)}

        # Test 2: Chat Completions API (for comparison)
        try:
            start_time = time.time()

            chat_result = self.client.chat.completions.create(
                model="gpt-4.5",  # Use non-reasoning model for fair comparison
                messages=[
                    {
                        "role": "system",
                        "content": "You are an expert cannabis compliance consultant for EQ12.",
                    },
                    {"role": "user", "content": cannabis_prompt},
                ],
            )

            chat_time = time.time() - start_time

            results["chat_completions"] = {
                "response": chat_result,
                "execution_time": chat_time,
                "token_count": chat_result.usage.total_tokens,
                "output_length": len(chat_result.choices[0].message.content),
            }

            print("✅ Chat Completions API completed")

        except Exception as e:
            logger.error(f"Chat Completions API error: {e}")
            results["chat_completions"] = {"error": str(e)}

        return results

    def analyze_results(self, results: dict[str, Any]):
        """
        Analyze and display comparison results
        """
        print("\n📊 API COMPARISON RESULTS")
        print("=" * 60)

        if "responses_api" in results and "error" not in results["responses_api"]:
            resp_data = results["responses_api"]
            print("\n🎯 RESPONSES API (GPT-5 with Reasoning):")
            print(f"   Execution Time: {resp_data['execution_time']:.2f}s")
            print(f"   Total Tokens: {resp_data['token_count']:,}")
            print(f"   Reasoning Tokens: {resp_data['reasoning_tokens']:,}")
            print(f"   Output Length: {resp_data['output_length']:,} chars")

        if "chat_completions" in results and "error" not in results["chat_completions"]:
            chat_data = results["chat_completions"]
            print("\n💬 CHAT COMPLETIONS API (GPT-4.5):")
            print(f"   Execution Time: {chat_data['execution_time']:.2f}s")
            print(f"   Total Tokens: {chat_data['token_count']:,}")
            print(f"   Output Length: {chat_data['output_length']:,} chars")

        # Calculate performance metrics
        if (
            "responses_api" in results
            and "chat_completions" in results
            and "error" not in results["responses_api"]
            and "error" not in results["chat_completions"]
        ):
            resp_data = results["responses_api"]
            chat_data = results["chat_completions"]

            print("\n📈 PERFORMANCE ANALYSIS:")

            # Quality comparison (rough estimate based on output length)
            quality_ratio = resp_data["output_length"] / chat_data["output_length"]
            print(f"   Output Quality Ratio: {quality_ratio:.2f}x")

            # Token efficiency
            token_efficiency = resp_data["token_count"] / chat_data["token_count"]
            print(f"   Token Usage Ratio: {token_efficiency:.2f}x")

            # Time comparison
            time_ratio = resp_data["execution_time"] / chat_data["execution_time"]
            print(f"   Time Ratio: {time_ratio:.2f}x")

            print("\n🧠 REASONING ADVANTAGE:")
            print(f"   Reasoning Tokens Used: {resp_data['reasoning_tokens']:,}")
            print("   Deep Analysis Capability: Enhanced")
            print("   Multi-step Problem Solving: Optimized")


# Run the API comparison
demo = ResponsesAPIDemo()
comparison_results = demo.compare_apis("cannabis compliance analysis")
demo.analyze_results(comparison_results)
```

## 4. Reasoning Effort Configuration

Configure the reasoning effort parameter (low, medium, high) and demonstrate how it affects token usage and response quality.

```python
# Reasoning effort configuration and analysis
class ReasoningEffortAnalyzer:
    """
    Analyze the impact of different reasoning effort levels
    """

    def __init__(self):
        self.client = client
        self.effort_levels = ["low", "medium", "high"]

    def test_reasoning_efforts(self) -> dict[str, Any]:
        """
        Test all reasoning effort levels on a complex EQ12 problem
        """

        # Complex credit assessment problem for EQ12
        credit_problem = """
        EQ12 Credit Assessment Challenge:

        A client applying for a $500,000 business loan presents:
        - Credit score: 720 (personal), 680 (business)
        - Annual revenue: $2.3M (growing 15% YoY for 3 years)
        - Debt-to-income ratio: 0.35
        - Cash flow: $180K positive monthly
        - Industry: Cannabis cultivation (legal state)
        - Collateral: $800K in equipment and real estate
        - 5 years in business, 2 years at current location

        Additional factors:
        - Recent regulatory changes in their state
        - Seasonal revenue fluctuations (40% variance)
        - Plans to expand to 2 new states in 18 months
        - Strong management team with industry experience
        - Banking relationship challenges due to industry

        Provide a comprehensive risk assessment with:
        1. Probability of default calculation
        2. Recommended loan terms and conditions
        3. Risk mitigation strategies
        4. Industry-specific considerations
        5. Regulatory compliance requirements

        Show your reasoning process step by step.
        """

        results = {}

        print("🎚️ TESTING REASONING EFFORT LEVELS")
        print("=" * 50)

        for effort in self.effort_levels:
            print(f"\n🧠 Testing effort level: {effort.upper()}")

            try:
                start_time = time.time()

                response = self.client.responses.create(
                    model="gpt-5",
                    reasoning={"effort": effort},
                    instructions="You are a senior credit analyst at EQ12 with expertise in cannabis industry lending.",
                    input=[{"role": "user", "content": credit_problem}],
                )

                execution_time = time.time() - start_time

                # Extract key metrics
                usage = response.usage
                reasoning_tokens = usage.output_tokens_details.get("reasoning_tokens", 0)

                results[effort] = {
                    "response": response,
                    "execution_time": execution_time,
                    "total_tokens": usage.total_tokens,
                    "input_tokens": usage.input_tokens,
                    "output_tokens": usage.output_tokens,
                    "reasoning_tokens": reasoning_tokens,
                    "output_length": len(response.output_text),
                    "status": response.status,
                }

                print(f"   ✅ Completed in {execution_time:.2f}s")
                print(f"   📊 Total tokens: {usage.total_tokens:,}")
                print(f"   🧮 Reasoning tokens: {reasoning_tokens:,}")
                print(f"   📝 Output length: {len(response.output_text):,} chars")

            except Exception as e:
                logger.error(f"Error with effort level {effort}: {e}")
                results[effort] = {"error": str(e)}

        return results

    def analyze_effort_impact(self, results: dict[str, Any]):
        """
        Analyze the impact of different effort levels
        """
        print("\n📈 REASONING EFFORT ANALYSIS")
        print("=" * 60)

        # Create comparison table
        print(
            f"\n{'Effort Level':<12} {'Time (s)':<10} {'Total Tokens':<15} {'Reasoning':<12} {'Output Chars':<12}"
        )
        print("-" * 65)

        valid_results = {}

        for effort in self.effort_levels:
            if effort in results and "error" not in results[effort]:
                data = results[effort]
                valid_results[effort] = data

                print(
                    f"{effort.capitalize():<12} "
                    f"{data['execution_time']:<10.2f} "
                    f"{data['total_tokens']:<15,} "
                    f"{data['reasoning_tokens']:<12,} "
                    f"{data['output_length']:<12,}"
                )

        # Performance analysis
        if len(valid_results) >= 2:
            print("\n🔍 PERFORMANCE INSIGHTS:")

            # Compare low vs high effort
            if "low" in valid_results and "high" in valid_results:
                low_data = valid_results["low"]
                high_data = valid_results["high"]

                token_increase = (high_data["total_tokens"] / low_data["total_tokens"] - 1) * 100
                reasoning_increase = (
                    high_data["reasoning_tokens"] / max(low_data["reasoning_tokens"], 1) - 1
                ) * 100
                quality_increase = (
                    high_data["output_length"] / low_data["output_length"] - 1
                ) * 100
                time_increase = (high_data["execution_time"] / low_data["execution_time"] - 1) * 100

                print(f"   💰 Token cost increase (low→high): +{token_increase:.1f}%")
                print(f"   🧠 Reasoning tokens increase: +{reasoning_increase:.1f}%")
                print(f"   📊 Output quality increase: +{quality_increase:.1f}%")
                print(f"   ⏱️ Processing time increase: +{time_increase:.1f}%")

        # Recommendations
        print("\n💡 EQ12 RECOMMENDATIONS:")
        print("   🎯 Low effort: Quick analysis, routine assessments")
        print("   ⚖️ Medium effort: Standard credit evaluations (recommended)")
        print("   🔬 High effort: Complex cases, regulatory analysis")

        return valid_results


# Run reasoning effort analysis
effort_analyzer = ReasoningEffortAnalyzer()
effort_results = effort_analyzer.test_reasoning_efforts()
analysis = effort_analyzer.analyze_effort_impact(effort_results)
```

## 5. Context Window Management

Implement proper context window management, reserving space for reasoning tokens and handling context limits effectively.

```python
# Context window management for reasoning models
class ContextWindowManager:
    """
    Manage context window efficiently for reasoning models
    """

    def __init__(self):
        self.client = client
        # GPT-5 context window (approximate)
        self.max_context_tokens = 200000  # Adjust based on actual model limits
        self.recommended_reasoning_buffer = 25000  # OpenAI recommendation

    def calculate_safe_context(self, input_text: str) -> dict[str, int]:
        """
        Calculate safe context limits for reasoning models
        """
        # Rough token estimation (1 token ≈ 4 characters)
        estimated_input_tokens = len(input_text) // 4

        # Reserve space for reasoning and output
        available_for_output = (
            self.max_context_tokens - estimated_input_tokens - self.recommended_reasoning_buffer
        )

        return {
            "estimated_input_tokens": estimated_input_tokens,
            "max_context_tokens": self.max_context_tokens,
            "reasoning_buffer": self.recommended_reasoning_buffer,
            "available_for_output": max(0, available_for_output),
            "recommended_max_output": min(available_for_output, 50000),  # Conservative limit
        }

    def demonstrate_context_management(self):
        """
        Demonstrate proper context window management for EQ12 workflows
        """
        print("🗂️ CONTEXT WINDOW MANAGEMENT DEMO")
        print("=" * 50)

        # Simulate a large EQ12 governance automation context
        large_context = (
            """
        EQ12 Governance Automation Analysis:

        Corporate Structure Assessment for Multi-State Cannabis Operation:

        BACKGROUND:
        Our client operates cannabis facilities across 8 states with the following structure:
        - California: 3 cultivation facilities, 2 manufacturing, 5 dispensaries
        - Colorado: 2 cultivation facilities, 1 manufacturing, 3 dispensaries
        - Nevada: 1 cultivation facility, 1 manufacturing, 2 dispensaries
        - Oregon: 2 cultivation facilities, 4 dispensaries
        - Washington: 1 cultivation facility, 3 dispensaries
        - Illinois: 1 cultivation facility, 2 dispensaries
        - Michigan: 2 cultivation facilities, 1 manufacturing, 3 dispensaries
        - Arizona: 1 cultivation facility, 2 dispensaries

        CURRENT GOVERNANCE ISSUES:
        1. Inconsistent compliance tracking across states
        2. Fragmented financial reporting systems
        3. Varying employee management protocols
        4. Different inventory management systems
        5. State-specific licensing requirements not centrally managed
        6. Tax reporting complexities
        7. Interstate commerce compliance gaps
        8. Vendor management inconsistencies
        9. Quality control standard variations
        10. Security protocol differences

        REGULATORY FRAMEWORK ANALYSIS:
        Each state has unique requirements for:
        - Product testing and certification
        - Seed-to-sale tracking systems
        - Advertising and marketing restrictions
        - Security and surveillance requirements
        - Financial record keeping
        - Employee background checks
        - Product packaging and labeling
        - Transportation and distribution
        - Waste disposal protocols
        - Customer verification processes

        FINANCIAL COMPLEXITY:
        - 280E tax implications across all states
        - State-specific tax rates and structures
        - Banking compliance challenges
        - Cash management protocols
        - Audit requirements by state
        - Transfer pricing between entities
        - Cost allocation methodologies
        - Revenue recognition standards

        OPERATIONAL CHALLENGES:
        - Supply chain coordination
        - Quality assurance standardization
        - Employee training consistency
        - Technology integration requirements
        - Data security and privacy compliance
        - Customer experience standardization
        - Inventory optimization across facilities
        - Equipment maintenance scheduling
        - Vendor qualification processes
        - Risk management protocols

        STRATEGIC OBJECTIVES:
        1. Implement unified governance framework
        2. Standardize compliance monitoring
        3. Optimize operational efficiency
        4. Ensure regulatory adherence
        5. Minimize compliance risks
        6. Streamline reporting processes
        7. Enhance data visibility
        8. Improve cost management
        9. Support expansion plans
        10. Maintain competitive advantage

        Please provide a comprehensive governance automation roadmap that addresses all these challenges while ensuring compliance across all jurisdictions.
        """
            * 3
        )  # Triple the content to test context limits

        # Calculate context requirements
        context_analysis = self.calculate_safe_context(large_context)

        print("📊 CONTEXT ANALYSIS:")
        print(f"   Input tokens (estimated): {context_analysis['estimated_input_tokens']:,}")
        print(f"   Max context window: {context_analysis['max_context_tokens']:,}")
        print(f"   Reasoning buffer: {context_analysis['reasoning_buffer']:,}")
        print(f"   Available for output: {context_analysis['available_for_output']:,}")
        print(f"   Recommended max output: {context_analysis['recommended_max_output']:,}")

        # Test with proper context management
        try:
            print("\n🎯 EXECUTING WITH CONTEXT MANAGEMENT...")

            response = self.client.responses.create(
                model="gpt-5",
                reasoning={"effort": "medium"},
                instructions="You are EQ12's senior governance consultant specializing in cannabis industry compliance.",
                input=[{"role": "user", "content": large_context}],
                max_output_tokens=context_analysis["recommended_max_output"],
            )

            print("✅ SUCCESS with context management!")
            print(f"   Status: {response.status}")
            print(f"   Actual tokens used: {response.usage.total_tokens:,}")
            print(
                f"   Reasoning tokens: {response.usage.output_tokens_details.get('reasoning_tokens', 0):,}"
            )
            print(f"   Output tokens: {response.usage.output_tokens:,}")

            # Check if response was truncated
            if response.status == "incomplete":
                print(f"⚠️ Response was incomplete due to: {response.incomplete_details.reason}")
                if response.incomplete_details.reason == "max_output_tokens":
                    print("💡 Consider increasing max_output_tokens or reducing input size")

            # Display first part of response
            print("\n📋 RESPONSE PREVIEW:")
            print(
                response.output_text[:500] + "..."
                if len(response.output_text) > 500
                else response.output_text
            )

            return response

        except Exception as e:
            logger.error(f"Context management error: {e}")
            print(f"❌ ERROR: {e}")
            return None

    def demonstrate_chunking_strategy(self, large_text: str) -> list[str]:
        """
        Demonstrate text chunking for very large contexts
        """
        print("\n📦 CHUNKING STRATEGY DEMO")
        print("=" * 40)

        # Calculate chunk size (leave room for reasoning and output)
        max_chunk_tokens = 50000  # Conservative chunk size
        chunk_chars = max_chunk_tokens * 4  # Rough character estimate

        chunks = []
        text_length = len(large_text)

        for i in range(0, text_length, chunk_chars):
            chunk = large_text[i : i + chunk_chars]
            chunks.append(chunk)

        print("📊 CHUNKING ANALYSIS:")
        print(f"   Total text length: {text_length:,} characters")
        print(f"   Chunk size: {chunk_chars:,} characters")
        print(f"   Number of chunks: {len(chunks)}")
        print("   Processing strategy: Sequential analysis with context linking")

        return chunks


# Demonstrate context window management
context_manager = ContextWindowManager()
governance_response = context_manager.demonstrate_context_management()

# Example of chunking for very large contexts
sample_large_text = "This is a very large document..." * 1000
chunks = context_manager.demonstrate_chunking_strategy(sample_large_text)
```

## 6. Cost Control with Token Limits

Use max_output_tokens parameter to control costs and handle incomplete responses due to token limits.

```python
# Cost control and token limit management
class CostControlManager:
    """
    Manage costs through strategic token limit configuration
    """

    def __init__(self):
        self.client = client
        # Example pricing (adjust based on actual OpenAI pricing)
        self.pricing = {
            "gpt-5": {
                "input_tokens_per_1k": 0.060,
                "output_tokens_per_1k": 0.240,
                "reasoning_tokens_per_1k": 0.240,  # Same as output tokens
            }
        }

    def calculate_cost(self, usage: dict[str, int], model: str = "gpt-5") -> dict[str, float]:
        """
        Calculate the cost of a request
        """
        if model not in self.pricing:
            return {"error": f"Pricing not available for {model}"}

        pricing = self.pricing[model]

        input_cost = (usage["input_tokens"] / 1000) * pricing["input_tokens_per_1k"]
        output_cost = (usage["output_tokens"] / 1000) * pricing["output_tokens_per_1k"]

        # Reasoning tokens are billed as output tokens
        reasoning_tokens = usage.get("reasoning_tokens", 0)
        reasoning_cost = (reasoning_tokens / 1000) * pricing["reasoning_tokens_per_1k"]

        total_cost = input_cost + output_cost

        return {
            "input_cost": input_cost,
            "output_cost": output_cost,
            "reasoning_cost": reasoning_cost,
            "total_cost": total_cost,
            "breakdown": {
                "input_tokens": usage["input_tokens"],
                "output_tokens": usage["output_tokens"],
                "reasoning_tokens": reasoning_tokens,
                "total_tokens": usage["total_tokens"],
            },
        }

    def test_cost_control_strategies(self):
        """
        Test different cost control strategies for EQ12 workflows
        """
        print("💰 COST CONTROL STRATEGIES")
        print("=" * 50)

        # EQ12 code generation task
        code_generation_task = """
        EQ12 Code Generation Request:

        Create a comprehensive Python module for EQ12's automated trading system with the following requirements:

        1. Real-time odds data processing from multiple sportsbooks
        2. Advanced arbitrage opportunity detection algorithms
        3. Risk management and position sizing calculations
        4. Portfolio optimization using modern portfolio theory
        5. Machine learning-based prediction models
        6. Automated bet placement with multiple APIs
        7. Comprehensive logging and monitoring system
        8. Database integration with PostgreSQL
        9. REST API for external integrations
        10. Unit tests with 90%+ coverage
        11. Performance optimization and caching
        12. Error handling and retry mechanisms
        13. Configuration management system
        14. Security features and authentication
        15. Real-time dashboard updates via WebSocket

        Include detailed documentation, type hints, and follow EQ12 coding standards.
        Generate production-ready code with comprehensive error handling.
        """

        # Test different token limits
        token_limits = [5000, 15000, 30000, 50000]
        results = {}

        for limit in token_limits:
            print(f"\n🎯 Testing with max_output_tokens: {limit:,}")

            try:
                response = self.client.responses.create(
                    model="gpt-5",
                    reasoning={"effort": "medium"},
                    instructions="You are EQ12's senior software architect. Generate production-quality code.",
                    input=[{"role": "user", "content": code_generation_task}],
                    max_output_tokens=limit,
                )

                # Calculate costs
                usage_dict = {
                    "input_tokens": response.usage.input_tokens,
                    "output_tokens": response.usage.output_tokens,
                    "reasoning_tokens": response.usage.output_tokens_details.get(
                        "reasoning_tokens", 0
                    ),
                    "total_tokens": response.usage.total_tokens,
                }

                cost_analysis = self.calculate_cost(usage_dict)

                results[limit] = {
                    "response": response,
                    "usage": usage_dict,
                    "cost": cost_analysis,
                    "status": response.status,
                    "output_length": len(response.output_text),
                }

                print(f"   Status: {response.status}")
                print(f"   Total tokens: {usage_dict['total_tokens']:,}")
                print(f"   Reasoning tokens: {usage_dict['reasoning_tokens']:,}")
                print(f"   Estimated cost: ${cost_analysis['total_cost']:.4f}")
                print(f"   Output length: {len(response.output_text):,} chars")

                if response.status == "incomplete":
                    print(f"   ⚠️ Incomplete: {response.incomplete_details.reason}")

            except Exception as e:
                logger.error(f"Error with token limit {limit}: {e}")
                results[limit] = {"error": str(e)}

        return results

    def analyze_cost_effectiveness(self, results: dict[int, dict]):
        """
        Analyze cost-effectiveness of different token limits
        """
        print("\n📊 COST-EFFECTIVENESS ANALYSIS")
        print("=" * 60)

        # Create analysis table
        print(
            f"{'Token Limit':<12} {'Status':<12} {'Total Cost':<12} {'Cost/Char':<12} {'Completion':<12}"
        )
        print("-" * 65)

        for limit, data in results.items():
            if "error" not in data:
                status = data["status"]
                cost = data["cost"]["total_cost"]
                output_length = data["output_length"]
                cost_per_char = cost / max(output_length, 1) * 1000  # Cost per 1000 chars
                completion = "Complete" if status == "completed" else "Incomplete"

                print(
                    f"{limit:<12,} {status:<12} ${cost:<11.4f} ${cost_per_char:<11.4f} {completion:<12}"
                )

        # Cost optimization recommendations
        print("\n💡 EQ12 COST OPTIMIZATION RECOMMENDATIONS:")

        # Find the sweet spot
        complete_results = {
            k: v for k, v in results.items() if "error" not in v and v["status"] == "completed"
        }

        if complete_results:
            # Find most cost-effective complete result
            best_value = min(
                complete_results.items(),
                key=lambda x: x[1]["cost"]["total_cost"] / x[1]["output_length"],
            )

            print(f"   🎯 Optimal token limit: {best_value[0]:,}")
            print(f"   💰 Cost: ${best_value[1]['cost']['total_cost']:.4f}")
            print("   📊 Output quality: Complete response")

        print("\n📋 STRATEGIC RECOMMENDATIONS:")
        print("   🔹 For quick analysis: 5,000-10,000 tokens")
        print("   🔹 For detailed reports: 15,000-25,000 tokens")
        print("   🔹 For code generation: 30,000-50,000 tokens")
        print("   🔹 For comprehensive analysis: 50,000+ tokens")

        return results

    def demonstrate_progressive_cost_control(self):
        """
        Demonstrate progressive cost control strategy
        """
        print("\n🎚️ PROGRESSIVE COST CONTROL STRATEGY")
        print("=" * 50)

        # Start with minimal tokens and increase if needed
        strategy_prompt = """
        EQ12 Cannabis Market Analysis:
        Analyze the competitive landscape for cannabis retail in California, focusing on:
        1. Market size and growth projections
        2. Key competitors and market share
        3. Pricing strategies and margins
        4. Regulatory environment impact
        5. Consumer trends and preferences
        Provide actionable insights for market entry strategy.
        """

        progressive_limits = [3000, 8000, 15000, 25000]

        for i, limit in enumerate(progressive_limits):
            print(f"\n📈 Step {i + 1}: Testing {limit:,} tokens")

            try:
                response = self.client.responses.create(
                    model="gpt-5",
                    reasoning={"effort": "low" if limit < 10000 else "medium"},
                    input=[{"role": "user", "content": strategy_prompt}],
                    max_output_tokens=limit,
                )

                print(f"   Status: {response.status}")
                print(f"   Tokens used: {response.usage.total_tokens:,}")

                if response.status == "completed":
                    print(f"   ✅ Sufficient at {limit:,} tokens")
                    break
                print("   ⚠️ Incomplete, trying higher limit...")

            except Exception as e:
                print(f"   ❌ Error: {e}")


# Run cost control analysis
cost_manager = CostControlManager()
cost_results = cost_manager.test_cost_control_strategies()
cost_analysis = cost_manager.analyze_cost_effectiveness(cost_results)

# Demonstrate progressive strategy
cost_manager.demonstrate_progressive_cost_control()
```

## 8. Error Handling and Status Checking

Implement robust error handling for incomplete responses and check response status for proper application flow.

```python
# Comprehensive error handling for reasoning models
class ReasoningErrorHandler:
    """
    Robust error handling and status management for reasoning models
    """

    def __init__(self):
        self.client = client
        self.retry_config = {
            "max_retries": 3,
            "base_delay": 1.0,
            "max_delay": 30.0,
            "backoff_factor": 2.0,
        }

    def handle_response_status(self, response) -> dict[str, Any]:
        """
        Comprehensive response status handling
        """
        status_info = {
            "status": response.status,
            "response_id": response.id,
            "is_complete": response.status == "completed",
            "has_output": hasattr(response, "output_text") and response.output_text,
            "token_usage": (
                response.usage._asdict()
                if hasattr(response.usage, "_asdict")
                else vars(response.usage)
            ),
        }

        if response.status == "incomplete":
            status_info.update(
                {
                    "incomplete_reason": response.incomplete_details.reason,
                    "recovery_strategy": self._get_recovery_strategy(
                        response.incomplete_details.reason
                    ),
                }
            )

        return status_info

    def _get_recovery_strategy(self, reason: str) -> str:
        """
        Determine recovery strategy based on incomplete reason
        """
        strategies = {
            "max_output_tokens": "Increase max_output_tokens or reduce input complexity",
            "content_filter": "Modify content to comply with usage policies",
            "context_length": "Reduce input size or use chunking strategy",
            "timeout": "Retry with simpler reasoning effort or smaller scope",
        }
        return strategies.get(reason, f"Unknown reason: {reason}. Contact support.")

    def demonstrate_error_scenarios(self):
        """
        Demonstrate various error scenarios and handling strategies
        """
        print("🛡️ ERROR HANDLING DEMONSTRATIONS")
        print("=" * 50)

        scenarios = [
            {
                "name": "Token Limit Exceeded",
                "setup": self._test_token_limit_error,
                "description": "Request exceeds max_output_tokens limit",
            },
            {
                "name": "Context Window Full",
                "setup": self._test_context_window_error,
                "description": "Input + reasoning tokens exceed context window",
            },
            {
                "name": "API Rate Limiting",
                "setup": self._test_rate_limit_error,
                "description": "Too many requests in short time period",
            },
            {
                "name": "Invalid Model Parameters",
                "setup": self._test_invalid_parameters,
                "description": "Invalid configuration parameters",
            },
        ]

        results = {}

        for scenario in scenarios:
            print(f"\n🎯 Testing: {scenario['name']}")
            print(f"   Description: {scenario['description']}")

            try:
                result = scenario["setup"]()
                results[scenario["name"]] = result
                print("   ✅ Scenario completed")

            except Exception as e:
                print(f"   ❌ Scenario failed: {e}")
                results[scenario["name"]] = {"error": str(e)}

        return results

    def _test_token_limit_error(self):
        """Test token limit exceeded scenario"""

        # Create a request that will exceed token limits
        complex_task = (
            """
        Generate a complete enterprise software architecture for EQ12 including:
        """
            + "Detailed requirements, " * 500
        )  # Force token limit issue

        try:
            response = self.client.responses.create(
                model="gpt-5",
                reasoning={"effort": "high"},
                input=[{"role": "user", "content": complex_task}],
                max_output_tokens=1000,  # Deliberately low
            )

            status_info = self.handle_response_status(response)

            if not status_info["is_complete"]:
                print("   ⚠️ Incomplete response detected")
                print(f"   📋 Reason: {status_info.get('incomplete_reason', 'Unknown')}")
                print(f"   🔧 Recovery: {status_info.get('recovery_strategy', 'None')}")

                # Attempt recovery
                if status_info.get("incomplete_reason") == "max_output_tokens":
                    print("   🔄 Attempting recovery with increased token limit...")

                    recovery_response = self.client.responses.create(
                        model="gpt-5",
                        reasoning={"effort": "medium"},  # Reduce effort
                        input=[
                            {
                                "role": "user",
                                "content": "Provide a high-level architecture overview for EQ12 enterprise software.",
                            }
                        ],
                        max_output_tokens=5000,  # Increased limit
                    )

                    recovery_status = self.handle_response_status(recovery_response)
                    print(f"   ✅ Recovery successful: {recovery_status['is_complete']}")

            return status_info

        except Exception as e:
            return {"error": str(e)}

    def _test_context_window_error(self):
        """Test context window exceeded scenario"""

        # Create a very large context
        large_context = "EQ12 data analysis: " + "Sample data point, " * 10000

        try:
            response = self.client.responses.create(
                model="gpt-5",
                reasoning={"effort": "high"},
                input=[{"role": "user", "content": large_context}],
            )

            return self.handle_response_status(response)

        except Exception as e:
            print(f"   ⚠️ Context window error: {e}")

            # Recovery strategy: chunking
            print("   🔄 Implementing chunking strategy...")

            chunk_size = 1000
            chunk = large_context[:chunk_size]

            recovery_response = self.client.responses.create(
                model="gpt-5",
                reasoning={"effort": "low"},
                input=[{"role": "user", "content": f"Analyze this data sample: {chunk}"}],
            )

            return {
                "original_error": str(e),
                "recovery_status": self.handle_response_status(recovery_response),
            }

    def _test_rate_limit_error(self):
        """Test rate limiting scenario"""

        # Simulate rapid requests (be careful not to actually hit rate limits)
        print("   ℹ️ Simulating rate limit scenario (not actually triggering)")

        # In real scenarios, you would implement exponential backoff
        def exponential_backoff_retry(func, *args, **kwargs):
            """Implement exponential backoff retry logic"""

            for attempt in range(self.retry_config["max_retries"]):
                try:
                    return func(*args, **kwargs)

                except Exception as e:
                    if (
                        "rate_limit" in str(e).lower()
                        and attempt < self.retry_config["max_retries"] - 1
                    ):
                        delay = min(
                            self.retry_config["base_delay"]
                            * (self.retry_config["backoff_factor"] ** attempt),
                            self.retry_config["max_delay"],
                        )
                        print(
                            f"   ⏳ Rate limit hit, waiting {delay:.1f}s before retry {attempt + 1}"
                        )
                        time.sleep(delay)
                    else:
                        raise e

            return None

        # Example usage (simulated)
        return {
            "strategy": "Exponential backoff implemented",
            "max_retries": self.retry_config["max_retries"],
            "status": "Ready for rate limit handling",
        }

    def _test_invalid_parameters(self):
        """Test invalid parameter handling"""

        try:
            # This should fail with invalid reasoning effort
            response = self.client.responses.create(
                model="gpt-5",
                reasoning={"effort": "extreme"},  # Invalid effort level
                input=[{"role": "user", "content": "Simple test"}],
            )

            return self.handle_response_status(response)

        except Exception as e:
            print(f"   ⚠️ Parameter validation error: {e}")

            # Recovery with valid parameters
            print("   🔄 Retrying with valid parameters...")

            recovery_response = self.client.responses.create(
                model="gpt-5",
                reasoning={"effort": "medium"},  # Valid effort level
                input=[{"role": "user", "content": "Simple test"}],
            )

            return {
                "parameter_error": str(e),
                "recovery_status": self.handle_response_status(recovery_response),
            }

    def create_robust_reasoning_client(self):
        """
        Create a robust wrapper for reasoning model calls
        """

        class RobustReasoningClient:
            def __init__(self, error_handler):
                self.client = error_handler.client
                self.error_handler = error_handler

            def create_with_retry(self, **kwargs):
                """Create response with automatic retry and error handling"""

                for attempt in range(self.error_handler.retry_config["max_retries"]):
                    try:
                        response = self.client.responses.create(**kwargs)

                        # Check response status
                        status_info = self.error_handler.handle_response_status(response)

                        if not status_info["is_complete"]:
                            # Handle incomplete responses
                            if status_info.get("incomplete_reason") == "max_output_tokens":
                                # Increase token limit and retry
                                kwargs["max_output_tokens"] = (
                                    kwargs.get("max_output_tokens", 10000) * 2
                                )
                                continue

                        return response, status_info

                    except Exception as e:
                        if attempt < self.error_handler.retry_config["max_retries"] - 1:
                            delay = self.error_handler.retry_config["base_delay"] * (2**attempt)
                            print(f"Attempt {attempt + 1} failed: {e}. Retrying in {delay}s...")
                            time.sleep(delay)
                        else:
                            raise e

                raise Exception("Max retries exceeded")

        return RobustReasoningClient(self)

    def demonstrate_robust_client(self):
        """
        Demonstrate the robust client wrapper
        """
        print("\n🛡️ ROBUST CLIENT DEMONSTRATION")
        print("=" * 50)

        robust_client = self.create_robust_reasoning_client()

        try:
            response, status_info = robust_client.create_with_retry(
                model="gpt-5",
                reasoning={"effort": "medium"},
                input=[
                    {
                        "role": "user",
                        "content": "Analyze EQ12's market position in the cannabis industry and provide strategic recommendations.",
                    }
                ],
                max_output_tokens=5000,
            )

            print("✅ Robust client successful!")
            print(f"   Status: {status_info['status']}")
            print(f"   Complete: {status_info['is_complete']}")
            print(f"   Tokens used: {status_info['token_usage']['total_tokens']:,}")

            return response, status_info

        except Exception as e:
            print(f"❌ Robust client failed: {e}")
            return None, None


# Demonstrate error handling capabilities
error_handler = ReasoningErrorHandler()
error_scenarios = error_handler.demonstrate_error_scenarios()

# Show robust client
robust_response, robust_status = error_handler.demonstrate_robust_client()
```
