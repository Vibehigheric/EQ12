### 4. BankrollManager.vb - Kelly Criterion Implementation

**Risk Management Features:**
- **Kelly Criterion**: Optimal bet sizing based on edge and bankroll
- **Risk Limits**: Maximum bet percentages and daily limits
- **Performance Tracking**: ROI, Sharpe ratio, maximum drawdown
- **SQLite Integration**: Persistent bankroll history and analytics

**Mathematical Foundation:**
- Kelly Formula: f = (bp - q) / b
- Fractional Kelly for reduced variance
- Risk of ruin calculations
- Monte Carlo simulations for outcome modeling

```python
# Kelly Criterion and Risk Management Analysis


def kelly_criterion_analysis():
    """Analyze Kelly Criterion implementation and risk management"""

    def kelly_fraction(win_prob, odds_decimal):
        """Calculate Kelly fraction for optimal bet sizing"""
        b = odds_decimal - 1  # Net odds received
        p = win_prob  # Probability of winning
        q = 1 - win_prob  # Probability of losing

        if b <= 0 or p <= 0:
            return 0

        kelly_f = (b * p - q) / b
        return max(0, kelly_f)  # Never bet negative amounts

    def fractional_kelly(kelly_f, fraction=0.25):
        """Apply fractional Kelly to reduce variance"""
        return kelly_f * fraction

    def risk_of_ruin(win_prob, kelly_f, bankroll_units=100):
        """Calculate risk of ruin probability"""
        if kelly_f >= 1.0:
            return 1.0  # Certain ruin if betting 100%+

        # Simplified risk of ruin calculation
        edge = (2 * win_prob) - 1
        if edge <= 0:
            return 1.0

        # Approximation for risk of ruin
        ror = math.exp(-2 * edge * bankroll_units * kelly_f)
        return min(1.0, ror)

    # Example scenarios
    scenarios = [
        {"name": "Conservative Value Bet", "win_prob": 0.52, "decimal_odds": 2.10},
        {"name": "Strong Edge", "win_prob": 0.58, "decimal_odds": 1.90},
        {"name": "High Confidence", "win_prob": 0.65, "decimal_odds": 1.75},
        {"name": "Arbitrage (Risk-Free)", "win_prob": 1.00, "decimal_odds": 1.02},
    ]

    print("💰 KELLY CRITERION ANALYSIS:")
    print("\n   Scenario Analysis:")

    for scenario in scenarios:
        name = scenario["name"]
        p = scenario["win_prob"]
        odds = scenario["decimal_odds"]

        full_kelly = kelly_fraction(p, odds)
        frac_kelly = fractional_kelly(full_kelly, 0.25)
        ror = risk_of_ruin(p, frac_kelly)

        print(f"\n      {name}:")
        print(f"        Win Prob: {p * 100:.1f}%")
        print(f"        Decimal Odds: {odds:.2f}")
        print(f"        Full Kelly: {full_kelly * 100:.2f}%")
        print(f"        Fractional Kelly (25%): {frac_kelly * 100:.2f}%")
        print(f"        Risk of Ruin: {ror * 100:.4f}%")

    # Risk Management Parameters
    risk_params = {
        "Max_Kelly_Fraction": "25% of full Kelly",
        "Daily_Limit": "10% of bankroll",
        "Max_Single_Bet": "5% of bankroll",
        "Stop_Loss": "20% drawdown limit",
        "Min_Edge_Required": "2% expected value",
    }

    print("\n   Risk Management Parameters:")
    for param, value in risk_params.items():
        print(f"      🛡️ {param.replace('_', ' ')}: {value}")


kelly_criterion_analysis()
```

### Arbitrage Detection Algorithm
**Two-Way Markets:**
- Identify opposing outcomes across books
- Calculate implied probabilities
- Detect negative total probability (arbitrage)
- Compute optimal stake distribution

**Three-Way Markets:**
- Handle win/draw/win scenarios
- Multi-dimensional optimization
- Risk-adjusted profit calculations

```python
# Arbitrage Detection Algorithm
def arbitrage_analysis():
    """Analyze arbitrage detection algorithms and profit calculations"""

    def american_to_decimal(american_odds):
        """Convert American odds to decimal format"""
        if american_odds > 0:
            return (american_odds / 100) + 1
        return (100 / abs(american_odds)) + 1

    def calculate_arbitrage(odds_list):
        """Calculate arbitrage opportunity and optimal stakes"""
        # Convert to decimal if needed
        decimal_odds = [
            american_to_decimal(odds) if odds < 0 or odds > 2 else odds for odds in odds_list
        ]

        # Calculate implied probabilities
        implied_probs = [1 / odds for odds in decimal_odds]
        total_prob = sum(implied_probs)

        # Check for arbitrage (total probability < 1)
        if total_prob >= 1.0:
            return None  # No arbitrage

        # Calculate profit percentage
        profit_pct = (1 - total_prob) / total_prob * 100

        # Calculate optimal stakes for $1000 total
        total_stake = 1000
        stakes = []
        payouts = []

        for _i, (prob, odds) in enumerate(zip(implied_probs, decimal_odds, strict=False)):
            stake = (prob / total_prob) * total_stake
            payout = stake * odds
            stakes.append(stake)
            payouts.append(payout)

        guaranteed_profit = min(payouts) - total_stake

        return {
            "profit_percentage": profit_pct,
            "guaranteed_profit": guaranteed_profit,
            "stakes": stakes,
            "payouts": payouts,
            "total_probability": total_prob,
        }

    # Example arbitrage scenarios
    scenarios = [
        {
            "name": "MLB H2H Example",
            "game": "Yankees vs Red Sox",
            "book_a": "DraftKings",
            "odds_a": 1.95,  # Yankees
            "book_b": "FanDuel",
            "odds_b": 2.10,  # Red Sox
        },
        {
            "name": "NBA Strong Arbitrage",
            "game": "Lakers vs Warriors",
            "book_a": "BetMGM",
            "odds_a": 2.05,  # Lakers
            "book_b": "Caesars",
            "odds_b": 1.98,  # Warriors
        },
    ]

    print("⚡ ARBITRAGE DETECTION ANALYSIS:")

    for scenario in scenarios:
        odds_list = [scenario["odds_a"], scenario["odds_b"]]
        result = calculate_arbitrage(odds_list)

        print(f"\n   {scenario['name']}:")
        print(f"      Game: {scenario['game']}")
        print(f"      {scenario['book_a']}: {scenario['odds_a']} decimal")
        print(f"      {scenario['book_b']}: {scenario['odds_b']} decimal")

        if result:
            print("      ✅ ARBITRAGE DETECTED!")
            print(f"         Profit: {result['profit_percentage']:.3f}%")
            print(f"         Guaranteed Profit: ${result['guaranteed_profit']:.2f}")
            print("         Stake Distribution:")
            print(f"           {scenario['book_a']}: ${result['stakes'][0]:.2f}")
            print(f"           {scenario['book_b']}: ${result['stakes'][1]:.2f}")
        else:
            print("      ❌ No arbitrage opportunity")

    # Performance metrics
    detection_metrics = {
        "Scan_Frequency": "Every 30 seconds",
        "Processing_Time": "< 500ms per market",
        "Min_Profit_Threshold": "1% guaranteed profit",
        "Max_Stake_Per_Arb": "10% of bankroll",
        "Opportunity_Lifespan": "2-5 minutes average",
    }

    print("\n   Detection Performance Metrics:")
    for metric, value in detection_metrics.items():
        print(f"      📊 {metric.replace('_', ' ')}: {value}")


arbitrage_analysis()
```
